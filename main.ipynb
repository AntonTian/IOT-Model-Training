```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Configuration
days_to_simulate = 30
start_date = datetime.now() - timedelta(days=days_to_simulate)
intervals = days_to_simulate * 24 * 12  

# Initialize DataFrame
date_rng = pd.date_range(start=start_date, periods=intervals, freq='5min')
df = pd.DataFrame(date_rng, columns=['timestamp'])

current_capacity = 0.0
capacities = []
open_counts = []

# Generate Time-Series Data
for dt in df['timestamp']:
    hour = dt.hour
    
    # Define peak hours 
    if (11 <= hour <= 14) or (18 <= hour <= 20):
        activity_chance = 0.7  
        max_fill_jump = 6.0    
    # Define sleep hours 
    elif 0 <= hour <= 6:
        activity_chance = 0.05 
        max_fill_jump = 1.0
    # Normal hours
    else:
        activity_chance = 0.3
        max_fill_jump = 3.0

    # Simulate waste being deposited
    if np.random.rand() < activity_chance:
        fill_amount = np.random.uniform(0.5, max_fill_jump)
        current_capacity += fill_amount
        opens = int(np.random.uniform(1, 4)) 
    else:
        opens = 0
        
    if current_capacity >= 90.0:
        current_capacity = 0.0
        
    capacities.append(round(current_capacity, 1))
    open_counts.append(opens)

df['capacity_percentage'] = capacities
df['open_count'] = open_counts

# Feature Engineering
df['hour_of_day'] = df['timestamp'].dt.hour
df['day_of_week'] = df['timestamp'].dt.dayofweek

# Export to CSV
df.to_csv('mock_smartbin_data.csv', index=False)
print("Successfully generated mock_smartbin_data.csv with", len(df), "rows.")
```
